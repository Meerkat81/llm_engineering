# Project - Food Truck Menu Maker
The user will enter a cuisine type and get four different menu items with recipe. 


*Uses remote model via api*

```python
# imports
from dotenv import load_dotenv
import os
from openai import OpenAI
import gradio as gr
```

```python
# Initialization
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins with {openai_api_key[:8]}")
else:
    print("OpenAi key not found")

MODEL = "gpt-4.1-mini"

openai = OpenAI()
```

```python
system_message = "You are a master chef able to come up with fantastic dish 5 " \
"ingredient dishes that can be made and served from a food truck. you provide the recipe and name"
```

```python
def chat(message, history):
    history = [{"role": h["role"], "content":h["content"]} for h in history]
    messages = [{"role":"system", "content":system_message}] + history + [{"role":"user", "content":message}]
    response = openai.chat.completions.create(model=MODEL, messages=messages)
    return response.choices[0].message.content

gr.ChatInterface(fn=chat, type="messages").launch()
```
