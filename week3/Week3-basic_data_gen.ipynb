## Imports

```python
from openai import OpenAI
import json
import gradio as gr
```

```python
openai = OpenAI(base_url = 'http://localhost:11434/v1', api_key='ollama')
```

## Models

```python
MODEL = "llama3.2"
```

## Prompts

```python
user_prompt = "Generate 4 different names for pasta dishes for a food truck. There are to be 4 themes: spicy, cheesy, seafood, vegetarian. the names will be in the following json format 'theme':'name'"

system_prompt = "You are a master chef able to come up with fantastic names for dishes"
```

## Access Model

```python
def generate_data(user, system):
    response = openai.chat.completions.create(
    model=MODEL,
    messages=[
        { "role":"system", "content":system},
        {"role":"user", "content":user}
        ],
    response_format={"type":"json_object"}
    )
    result = json.loads(response.choices[0].message.content)
    return result
```

## Generate Data

```python
result = generate_data(user_prompt, system_prompt)
formatted_json_result = json.dumps(result, indent=4)
```
